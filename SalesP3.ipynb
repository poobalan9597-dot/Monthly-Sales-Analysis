```python
monthly_sales=df.groupby(["Month_Number","Month"])["Sales"].sum().reset_index()
monthly_sales.sort_values('Month_Number')
```

```python
plt.figure(figsize=(10,5))
plt.plot(monthly_sales["Month"],monthly_sales["Sales"],marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
prd_sales=df.groupby("Product")["Sales"].sum()
plt.figure(figsize=(8,5))
plt.bar(prd_sales.index,prd_sales.values)
plt.title("Product sales")
plt.xlabel("Product")
plt.ylabel("sales(in lakhs)")
plt.show()
```

```python
prd_quantity=df.groupby("Product")["Quantity"].sum()
plt.figure(figsize=(7,5))
plt.barh(prd_quantity.index,prd_quantity.values,color="b")
plt.title("Qunatity of Products")
plt.xlabel("Product")
plt.ylabel("Quantity")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Quantity"],df["Sales"],s=50,c=df["Sales"],cmap="viridis",alpha=0.9)
plt.colorbar(label="sales")
plt.title("Quantity vs Sales")
plt.xlabel("Qunatity")
plt.ylabel("Sales")
plt.show()
```

```python
a=df.groupby("Product")["Quantity"].sum()
colors=["r","b","g","y","Orange","purple","pink"]
explode=(0,0,0,0,0,0.2,0)
plt.figure(figsize=(6,6))
plt.pie(a.values,labels=a.index,autopct="%1.1f%%",startangle=90,colors=colors,explode=explode,shadow=True)
plt.title("Quantity of Product")
plt.show()
```

```python
plt.figure(figsize=(7,7))
Sales_Channel=df.groupby("Sales_Channel")["Sales"].sum()
plt.pie(Sales_Channel,labels=Sales_Channel.index,autopct='%1.1f%%',startangle=90)
plt.title("ONLINE VS STORE SALES")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Sales"],bins=15,color="r",edgecolor="b")
plt.title("Sales of Product")
plt.xlabel("Price")
plt.ylabel("sales")

plt.show()
```

```python
from matplotlib import markers
monthly_sales=df.groupby("Month")["Sales"].sum()
prd_sales=df.groupby("Product")["Sales"].sum()
prd_quantity=df.groupby("Product")["Quantity"].sum()
plt.figure(figsize=(12,8))

plt.subplot(2,2,1)
plt.plot(monthly_sales.index,monthly_sales.values,marker="o",color="b")
plt.title("Monthly sales Trend")
plt.xticks(rotation=90)


plt.subplot(2,2,2)
plt.bar(prd_sales.index,prd_sales.values)
plt.title("Sales by Product")
plt.xticks(rotation=45)


plt.subplot(2,2,3)
plt.scatter(df['Quantity'],df['Profit'],s=50,c=df["Sales"],cmap="viridis",alpha=0.6)
plt.colorbar(label="sales")
plt.title("Quantity of profit")


plt.subplot(2,2,4)
colors=["r","orange","g","b","y","purple","pink"]
plt.pie(prd_sales.values,labels=prd_sales.index,autopct='%1.1f%%',colors=colors,explode=explode)
plt.title("Product Percentage")
explode=(0.1,0,0,0,0,0,0)
plt.tight_layout()
plt.show()
```

```python
sales_table = pd.pivot_table(
    df,
    values="Sales",
    index="Month",
    columns="Product",
    aggfunc="sum",
    fill_value=0
)

month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

sales_table = sales_table.reindex(month_order)

print(sales_table)

```

```python
sales_table.plot(
    kind="bar",
    figsize=(14, 6),
    width=0.8
)

plt.title("Monthly Sales Comparison - All Products")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.legend(title="Products")
plt.tight_layout()
plt.show()

```
